```python
import pyarrow
import pandas as pd
import numpy as np
import pyodbc
from flask import Flask, render_template
```

```python
customers = pd.read_csv('customers.csv')
customers
```

```python
customers.drop(4,inplace=True)
customers
```

```python
customers['city'] = customers['city'].str.lower()
customers
```

```python
customers.info()
```

```python
customers.fillna('none', inplace=True)
customers
```

```python
customers['fullname'] = customers['first_name'] + ' ' + customers['last_name']
customers
```

```python
orders = pd.read_parquet('orders.parquet', engine ='pyarrow')
orders
```

```python
orders.drop_duplicates(inplace = True)
orders
```

```python
orders.info()
```

```python
orders.loc[5, 'customer_id'] = 1
orders
```

```python
url = 'https://raw.githubusercontent.com/MohammedHameds/test1455/refs/heads/main/products.csv'
products = pd.read_csv(url)
products
```

```python
products.duplicated().sum()
```

```python
products['category'] = products['category'].str.strip()
products['product_name'] = products['product_name'].str.strip()
products
```

```python
products.info()
```

```python
sales = orders.merge(customers, how='inner', on='customer_id').merge(products, how='inner', on='product_id')
sales
```

```python
sales["total_amount"] = sales["quantity"] * sales["unit_price"]
sales
```

```python
sales = sales[['order_id','fullname','product_name','quantity','unit_price','total_amount']]
sales.rename(columns={'fullname' : 'customer', 'product_name' : 'product'}, inplace = True)
sales
```

```python
conn = pyodbc.connect(
    "driver={ODBC Driver 17 for SQL Server};"
    "server=localhost\\SQLEXPRESS;"
    "database=master;"
    "trusted_connection=yes;"
)

cursor = conn.cursor()
```

```python
cursor.execute('create table sales (id int primary key , customer varchar(50) , product varchar(50) , quantity int, price int, total_amount int)')
cursor.commit()
```

```python
rows = list(sales.itertuples(index=False, name=None))
rows
```

```python
cursor.executemany('insert into sales values (?,?,?,?,?,?)', rows)
cursor.commit()
```

```python
app = Flask(__name__)

@app.route('/')
def home():
    df = pd.read_sql_query('select * from sales', conn)
    return render_template('home.html', table = df.to_html())


if __name__ == '__main__':
    app.run(port = 5000)
```
